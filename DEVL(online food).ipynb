```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV dataset
df = pd.read_csv("/home/pccoe/Downloads/online food delivery dataset.csv")

# Remove unwanted extra column
df = df.drop(columns=["Unnamed: 13"], errors="ignore")

# Display first five records
print("First five records:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
print(df.info())

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe(include="all"))

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check dimensions
print("\nDataset Shape:", df.shape)


# ------------------------------------------------
# 1. Line Plot
# ------------------------------------------------
plt.figure(figsize=(8, 5))

plt.plot(df.index, df["Age"], label="Age")
plt.plot(df.index, df["Family size"], label="Family Size")

plt.title("Line Plot - Age and Family Size")
plt.xlabel("Observation Number")
plt.ylabel("Value")

plt.legend()
plt.show()


# ------------------------------------------------
# 2. Bar Plot
# ------------------------------------------------
# Count customers according to occupation

occupation_count = df["Occupation"].value_counts()

plt.figure(figsize=(8, 5))
occupation_count.plot(kind="bar")

plt.title("Number of Customers by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------
# 3. Histogram
# ------------------------------------------------
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Histogram of Customer Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


# ------------------------------------------------
# 4. Scatter Plot
# ------------------------------------------------
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Family size",
    hue="Gender"
)

plt.title("Age vs Family Size")
plt.xlabel("Age")
plt.ylabel("Family Size")

plt.show()


# ------------------------------------------------
# 5. Pie Chart
# ------------------------------------------------
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Customers by Gender")

plt.show()


# ------------------------------------------------
# 6. Box Plot
# ------------------------------------------------
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Gender",
    y="Age"
)

plt.title("Box Plot of Age by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")

plt.show()


# ------------------------------------------------
# 7. Additional Seaborn Visualization
# ------------------------------------------------

sns.countplot(
    data=df,
    x="Feedback"
)

plt.title("Customer Feedback")
plt.xlabel("Feedback")
plt.ylabel("Number of Customers")

plt.show()
```
